```python
# E-commerce Delivery Network Analysis

#This notebook analyzes the delivery network using shortest path optimization.
```

```python
import networkx as nx
import pandas as pd

df = pd.read_csv("../data/network_data.csv")

G = nx.from_pandas_edgelist(df, source='source', target='target', edge_attr='time')

shortest_path = nx.shortest_path(G, source="Depot", target="Levent", weight='time')
shortest_time = nx.shortest_path_length(G, source="Depot", target="Levent", weight='time')

shortest_path, shortest_time
```

```python
## Interpretation

#The optimal route minimizes delivery time by passing through Atasehir and Sisli.
#This demonstrates efficient routing in urban logistics.
```
